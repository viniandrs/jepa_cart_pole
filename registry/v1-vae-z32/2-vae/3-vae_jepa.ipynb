# **JEPA Training - Latent Dynamics Predictor for CartPole**
---

This notebook trains the **predictor** of a Joint-Embedding Predictive Architecture (JEPA) world model on top of the VAE trained in `1-vae_training.ipynb`. Given the latent representation of the last frames and the action taken, it predicts the latent representation of the next frame.

## **Theory: Predicting in Representation Space**

A *generative* world model predicts the next observation $x_{t+1}$ pixel by pixel. A **JEPA** (LeCun, 2022; I-JEPA, Assran et al., 2023) instead predicts the next observation's **embedding**:

$$\hat{z}_{t+1} = f_\theta\big(z_{t-k+1}, \dots, z_t,\ a_t\big), \qquad z_t = \mathrm{Enc}(x_t)$$

and the loss is measured in the embedding space, not in pixels. The encoder can discard details that are irrelevant or unpredictable (textures, background), so the predictor only has to model what matters for the dynamics — here, the cart position and pole angle.

When the encoder is trained jointly with the predictor, JEPAs need extra machinery (an EMA target encoder, variance/covariance regularisation) to prevent **representation collapse** — mapping every frame to the same constant, which makes prediction trivially perfect. In this project the encoder is the **frozen, pretrained VAE**: its embeddings are fixed by the reconstruction objective, so collapse cannot happen and the predictor is trained with a plain regression loss. This is also the setting of World Models' memory (M) module, with a deterministic MLP instead of an MDN-RNN.

## **Loss Function**

Following World Models, the latent of every frame is **sampled** from the stored posterior on each batch, which acts as data augmentation and keeps the predictor in the region the decoder understands:

$$z_t = \mu_t + \sigma_t \odot \varepsilon, \quad \varepsilon \sim \mathcal{N}(0, I)$$

$$\mathcal{L}(\theta) = \frac{1}{B\,d} \sum_{i=1}^{B} \big\lVert f_\theta(z^{(i)}_{t-k+1:t},\ a^{(i)}_t) - z^{(i)}_{t+1} \big\rVert_2^2$$

The minimiser of this MSE is the conditional mean $\mathbb{E}[z_{t+1} \mid z_{t-k+1:t}, a_t]$, so sampling the target adds noise to the gradients but not bias. The training loss therefore includes the irreducible variance of the sampled target ($\approx \bar\sigma^2$, close to 1 for the VAE's unused dimensions), and **validation is measured on the means** $\mu$, which is the quantity we actually want to predict.

## **Why a Context of Several Latents?**

A single 64×64 frame shows *where* the cart and pole are, but not how fast they move. The CartPole state $(x, \dot{x}, \theta, \dot{\theta})$ is therefore only **partially observable** from one frame, and $(z_t, a_t)$ does not determine $z_{t+1}$. Feeding the last $k$ latents (`context_len`, default 2) lets the MLP estimate velocities by finite differences, exactly as frame stacking does in DQN.

## **Baseline**

Consecutive frames are very similar, so a small MSE alone says little. Every result is compared against the **copy-last baseline** $\hat{z}_{t+1} = z_t$: a useful predictor must beat it, i.e. it must actually model the effect of the dynamics and the action.

## **Setup and Imports**

```python
from jepacartpole.config import ConfigVAE, ConfigJEPA
from jepacartpole.vae import VAE
from jepacartpole.jepa import (
    JEPAPredictor,
    JEPADataset,
    train_jepa,
    evaluate_jepa,
    plot_jepa_training_curves,
    plot_predictions,
    plot_rollout_error,
    plot_rollout_frames
)
from jepacartpole.utils import (
    get_device,
    load_checkpoint,
    count_parameters,
    set_seed,
    get_checkpoint_dir,
    get_data_dir
)

import torch
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

# Set random seed for reproducibility
set_seed(42)

# Get device
DEVICE = get_device()
print(f"Device: {DEVICE}")
```

## **Configuration**
---

```python
# Model configuration
config = ConfigJEPA()
vae_config = ConfigVAE()
assert config.z_dim == vae_config.latent_dim, "ConfigJEPA.z_dim must match ConfigVAE.latent_dim"

# Latents produced by notebooks/1-data_prep/2-latent_encoding.ipynb
LATENT_PATH = get_data_dir('jepa') / 'vae_latents.h5'
# Frames, only used to display ground-truth images
FRAMES_PATH = get_data_dir('vae') / 'vae_data.h5'

print("JEPA Configuration:")
print(f"  Latent dimension: {config.z_dim}")
print(f"  Action dimension: {config.action_dim}")
print(f"  Context length: {config.context_len}")
print(f"  Hidden dimension: {config.hidden_dim}")
print(f"  Sample latents: {config.sample_latents}")
print(f"  Batch size: {config.batch_size}")
print(f"  Epochs: {config.epochs}")
print(f"  Learning rate: {config.learning_rate}")
print(f"  Weight decay: {config.weight_decay}")
print(f"  Validation fraction: {config.val_fraction}")
print()
print(f"Dataset: {LATENT_PATH}")
```

## **Model Architecture**
---

A 3-layer MLP over the flattened context and the action: `context_len·z_dim + action_dim → hidden → hidden → z_dim`.

```python
# Initialize model
model = JEPAPredictor(config).to(DEVICE)

# Display architecture
print("JEPA Predictor Architecture Summary:")
print("=" * 80)
print(model)
print()
print(f"Trainable parameters: {count_parameters(model):,}")
```

## **Training**
---

Episodes are split into training and validation sets (`val_fraction` of the episodes, split by episode so correlated frames never leak across sets). Only transitions whose context and target belong to the same sub-episode are used (see `2-latent_encoding.ipynb`). The whole latent dataset fits in RAM, so an epoch takes a few seconds on CPU.

The model is checkpointed to `checkpoints/jepa_latest.pth` after every epoch and to `checkpoints/jepa.pth` at the end.

```python
%%time

# Train the JEPA predictor
loss_history = train_jepa(
    model=model,
    config=config,
    h5_path=LATENT_PATH,
    device=DEVICE,
    # epochs=1,
    verbose=True
)
```

### **Training Curves**

The training loss is computed on sampled latents and includes the target's sampling variance, so it sits far above the validation loss (on $\mu$). What matters is the validation curve relative to the copy-last baseline.

```python
# Plot training curves
fig = plot_jepa_training_curves(loss_history)
plt.show()

print(f"\nTraining Summary:")
print(f"  Final val loss: {loss_history['epoch_val_losses'][-1]:.6f}")
print(f"  Best val loss: {loss_history['best_val_loss']:.6f}")
print(f"  Copy-last baseline: {loss_history['baseline_val_loss']:.6f}")
print(f"  Improvement over baseline: {loss_history['baseline_val_loss'] / loss_history['best_val_loss']:.1f}×")
print(f"  Total batches: {loss_history['total_batches']:,}")
```

## **Load Pre-trained Model**
---

If you've already trained the model, you can skip training and load the checkpoint directly. The VAE is loaded as well, to decode predicted latents into images.

```python
# Load checkpoints
checkpoint_path = get_checkpoint_dir() / 'jepa.pth'

if checkpoint_path.exists():
    model = JEPAPredictor(config).to(DEVICE)
    checkpoint = load_checkpoint(model, checkpoint_path, device=DEVICE)
    val_episodes = checkpoint['val_episodes']

    print(f"✓ Loaded model from: {checkpoint_path}")
    print(f"  Training epoch: {checkpoint.get('epoch', 'unknown')}")
    print(f"  Best val loss: {checkpoint.get('best_val_loss', 'unknown')}")
else:
    print(f"✗ Checkpoint not found at: {checkpoint_path}")
    print("  Train the model first (see Training section above)")

vae = VAE(vae_config).to(DEVICE)
load_checkpoint(vae, get_checkpoint_dir() / 'vae.pth', device=DEVICE)
vae.eval()

# Validation transitions (episodes never seen during training)
val_dataset = JEPADataset(LATENT_PATH, context_len=config.context_len, episodes=val_episodes)
val_loss, baseline_loss = evaluate_jepa(model, DataLoader(val_dataset, batch_size=1024), DEVICE)
print(f"\nValidation: {len(val_dataset):,} transitions")
print(f"  JEPA MSE:      {val_loss:.6f}")
print(f"  Copy-last MSE: {baseline_loss:.6f}")
```

## **One-Step Predictions**
---

The predicted latents can be turned back into images with the VAE decoder. The middle row (decoding the true $\mu_{t+1}$) is the best the predictor can achieve through the decoder; the bottom row should match it closely.

```python
fig = plot_predictions(
    vae=vae,
    predictor=model,
    dataset=val_dataset,
    frames_h5_path=FRAMES_PATH,
    n=5
)
plt.show()
```

## **Open-Loop Rollouts (Imagination)**
---

A world model is useful for planning only if it can predict several steps ahead **without** observing new frames. In an open-loop rollout the predictor's output is fed back as context:

$$\hat{z}_{t+h+1} = f_\theta(\hat{z}_{t+h-k+1:t+h},\ a_{t+h})$$

using the recorded actions. Errors compound over the horizon, so the gap to the copy-last baseline at long horizons measures how well the dynamics were learned. Horizons are limited to ~8 steps because, with random actions, the pole falls every ~10 steps (segments crossing a reset are excluded).

```python
fig = plot_rollout_error(
    predictor=model,
    dataset=val_dataset,
    horizon=8,
    n=500
)
plt.show()
```

```python
fig = plot_rollout_frames(
    vae=vae,
    predictor=model,
    dataset=val_dataset,
    frames_h5_path=FRAMES_PATH,
    horizon=8,
    seed=0
)
plt.show()
```

## **Analysis: Latent Dimensionality**
---

The VAE's KL term only pays for information above the floor $\tau \cdot d$, so the encoder is free to leave some latent dimensions at the prior $\mathcal{N}(0, 1)$. Those dimensions carry no information about the frame, yet the predictor still receives them as input and must predict them as targets. This section measures how many dimensions are actually used and how that affects JEPA training.

### **Latent Usage**

For each dimension of the validation targets we compute the average KL to the prior (same "active if KL > 0.05" criterion as `2-vae_evaluation.ipynb`) and the average posterior variance $\bar\sigma^2$.

When training on sampled latents, the target $z_{t+1} = \mu_{t+1} + \sigma_{t+1}\varepsilon$ contains noise that no predictor can anticipate, so the training MSE cannot go below the **noise floor** $\frac{1}{d}\sum_j \bar\sigma_j^2$ (plus the extra error caused by the noisy inputs).

```python
import numpy as np

mu_val = val_dataset.mu.reshape(-1, config.z_dim)
var_val = val_dataset.logvar.exp().reshape(-1, config.z_dim)

kl_per_dim = (0.5 * (mu_val.pow(2) + var_val - 1 - var_val.log())).mean(dim=0).numpy()
mean_sigma = var_val.sqrt().mean(dim=0).numpy()
active = kl_per_dim > 0.05
noise_floor = var_val.mean().item()

print(f"Active dimensions (KL > 0.05): {active.sum()} / {config.z_dim} -> {np.nonzero(active)[0].tolist()}")
print(f"Mean σ of active dims:   {mean_sigma[active].mean():.3f}")
if (~active).any():
    print(f"Mean σ of inactive dims: {mean_sigma[~active].mean():.3f}")
print()
print(f"Noise floor of sampled targets (mean σ²): {noise_floor:.4f}")
print(f"Final train loss (sampled z):             {checkpoint['epoch_train_losses'][-1]:.4f}")
print(f"Final val loss (μ):                       {checkpoint['epoch_val_losses'][-1]:.6f}")
```

### **Error by Dimension**

Validation MSE of the predictor and of the copy-last baseline for each latent dimension (on $\mu$). Dimensions that the VAE does not use are almost constant, so both models predict them trivially; the dynamics live entirely in the active dimensions.

```python
sq_err_model = torch.zeros(config.z_dim)
sq_err_base = torch.zeros(config.z_dim)
n = 0
model.eval()
with torch.no_grad():
    for mu_ctx, _, action, mu_next, _ in DataLoader(val_dataset, batch_size=4096):
        mu_ctx, action, mu_next = mu_ctx.to(DEVICE), action.to(DEVICE), mu_next.to(DEVICE)
        sq_err_model += (model(mu_ctx, action) - mu_next).pow(2).sum(dim=0).cpu()
        sq_err_base += (mu_ctx[:, -1] - mu_next).pow(2).sum(dim=0).cpu()
        n += len(mu_next)
mse_model = (sq_err_model / n).numpy()
mse_base = (sq_err_base / n).numpy()

dims = np.arange(config.z_dim)
width = 0.4
fig, ax = plt.subplots(figsize=(15, 4))
ax.bar(dims - width / 2, mse_base, width, color='gray', label='Copy-last baseline')
ax.bar(dims + width / 2, mse_model, width, label='JEPA predictor')
for d in np.nonzero(active)[0]:
    ax.axvspan(d - 0.5, d + 0.5, color='orange', alpha=0.15, linewidth=0)
ax.set_yscale('log')
ax.set_xticks(dims)
ax.set_xlabel('Latent dimension (shaded = active)', fontsize=12)
ax.set_ylabel('Validation MSE (μ)', fontsize=12)
ax.set_title('Prediction Error per Latent Dimension', fontsize=14, fontweight='bold')
ax.grid(True, axis='y', alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

print(f"Share of JEPA error in active dims:     {mse_model[active].sum() / mse_model.sum():.1%}")
print(f"Share of baseline error in active dims: {mse_base[active].sum() / mse_base.sum():.1%}")
```

### **Ablation: Sampled $z$ vs. $\mu$**

The same predictor is trained again with the other setting of `sample_latents` (same epochs, seed and split; checkpoints go to a temporary directory so `checkpoints/jepa.pth` is not overwritten). Both are evaluated on $\mu$.

```python
%%time
import tempfile
from dataclasses import replace

ablation_config = replace(config, sample_latents=not config.sample_latents)
with tempfile.TemporaryDirectory() as tmp_dir:
    ablation_history = train_jepa(
        model=JEPAPredictor(ablation_config).to(DEVICE),
        config=ablation_config,
        h5_path=LATENT_PATH,
        checkpoint_dir=tmp_dir,
        device=DEVICE,
        verbose=False
    )

runs = {config.sample_latents: checkpoint['epoch_val_losses'],
        ablation_config.sample_latents: ablation_history['epoch_val_losses']}

fig, ax = plt.subplots(figsize=(10, 5))
epochs = range(1, len(runs[True]) + 1)
ax.plot(epochs, runs[True], 'o-', linewidth=2, markersize=6, label='Trained on sampled z')
ax.plot(epochs, runs[False], 'o-', linewidth=2, markersize=6, color='orange', label='Trained on μ')
ax.axhline(checkpoint['baseline_val_loss'], linestyle='--', color='gray', label='Copy-last baseline')
ax.set_yscale('log')
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('Validation MSE (μ)', fontsize=12)
ax.set_title('Ablation: Training Targets', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

for name, losses in (('sampled z', runs[True]), ('μ', runs[False])):
    print(f"Best val MSE trained on {name:9s}: {min(losses):.6f}")
print(f"Copy-last baseline:              {checkpoint['baseline_val_loss']:.6f}")
```

### **Interpretation**

**Results for v1 (`latent_dim = 32`, `kl_tolerance = 0.5` → 16-nat KL floor):**

| | Value |
|---|---|
| Active latent dimensions | **4 / 32** (dims 1, 5, 27, 29) |
| Mean σ, active / inactive dims | 0.18 / 0.95 |
| Noise floor of sampled targets (mean σ²) | 0.808 |
| Final train loss (sampled z) | 0.809 |
| Best val MSE, trained on sampled z | 0.00061 |
| Best val MSE, trained on μ | 0.00032 |
| Copy-last baseline | 0.0052 |
| Share of JEPA val error in the 28 inactive dims | ~30% |

1. **The VAE uses only 4 of its 32 dimensions.** With a 16-nat free-bits budget, the encoder concentrates the information about the cart position and pole angle in a few very precise dimensions (σ ≈ 0.18) and leaves the other 28 at the prior (σ ≈ 0.95). This matches `2-vae_evaluation.ipynb`: CartPole has only two visible degrees of freedom, so 32 dimensions are far more than needed.
2. **Sampled-z training is dominated by noise.** The training loss (0.809) is essentially the noise floor (0.808): 99.9% of what the model sees as "error" is the unpredictable sampling noise of the 28 unused dimensions. The gradient signal of the dynamics is buried in it, which is why training on μ reaches a **~2× lower** validation error (0.00032 vs. 0.00061) with the same model, data and epochs.
3. **The unused dimensions still cost accuracy.** They are nearly constant, so the copy-last baseline predicts them almost perfectly, but the JEPA predictor is *worse* than the baseline on most of them and they account for ~30% of its validation error: the predictor has to learn to ignore 28 noisy inputs and output 28 constants.
4. **The predictor does learn the dynamics.** In the active dimensions its error is ~10× below the baseline, the decoded predictions match the targets, and open-loop rollouts stay well below the baseline for 8 steps.

**Conclusion:** the world model works, but the latent space is oversized. A VAE with `latent_dim = 4` (the number of active dimensions found here) should remove the noise-dominated dimensions, make sampled-latent training meaningful, and make the predictor's job smaller.
