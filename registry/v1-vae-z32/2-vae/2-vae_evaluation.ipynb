# **VAE Evaluation**
---

Debug notebook to check whether the trained VAE (`checkpoints/vae.pth`) produces representations that are useful for the JEPA world model:

1. **Training history**: did training converge?
2. **Reconstruction quality**: can the decoder rebuild the frames?
3. **Posterior collapse**: how many latent dimensions carry information?
4. **Latent traversals**: what does each active dimension encode?
5. **Physical state probing**: can cart position, cart velocity, pole angle and angular velocity be recovered from the latents?

```python
import numpy as np
import torch
import matplotlib.pyplot as plt

from jepacartpole.config import ConfigVAE
from jepacartpole.vae import VAE, CartPoleVAEDataset, plot_training_curves, plot_reconstructions, get_sample_images
from jepacartpole.data_prep import setup_pygame_headless, create_cartpole_env
from jepacartpole.utils import get_device, load_checkpoint, get_checkpoint_dir, get_data_dir, set_seed

set_seed(42)
DEVICE = get_device()
H5_PATH = get_data_dir('vae') / 'vae_data.h5'
CHECKPOINT_PATH = get_checkpoint_dir() / 'vae.pth'

model = VAE(ConfigVAE())
checkpoint = load_checkpoint(model, CHECKPOINT_PATH, device=DEVICE)
model.eval()

config = checkpoint['config']
print(f"Device: {DEVICE}")
print(f"Loaded {CHECKPOINT_PATH} (epoch {checkpoint['epoch'] + 1})")
print(f"Config: {config}")
```

## **1. Training History**
---

The loss is `recon + β · max(KL, kl_tolerance · latent_dim)`. With `kl_tolerance = 0.5` and 32 latent dimensions, the KL floor is **16 nats**: once KL drops below it, the KL term stops being optimized and training focuses on reconstruction only.

```python
fig = plot_training_curves(checkpoint)
plt.show()

kl_floor = config.kl_tolerance * config.latent_dim
print(f"Final recon loss: {checkpoint['epoch_recon_losses'][-1]:.4f} (sum of squared errors per image)")
print(f"Final KL:         {checkpoint['epoch_kl_losses'][-1]:.4f} (floor = {kl_floor:.0f})")
```

## **2. Reconstruction Quality**
---

```python
sample_images = get_sample_images(H5_PATH, n=8, seed=0)
fig = plot_reconstructions(model, sample_images.to(DEVICE), n=8, figsize=(16, 4))
plt.show()

# Reconstruction metrics on a random subset of the dataset
dataset = CartPoleVAEDataset(H5_PATH, in_memory=False)
rng = np.random.default_rng(0)
indices = np.sort(rng.choice(len(dataset), size=5000, replace=False))
images = torch.stack([dataset[i][0] for i in indices])
dataset.close()

with torch.no_grad():
    mu, logvar = model.encode(images.to(DEVICE))
    recon = model.decode(mu)

mse = ((recon.cpu() - images) ** 2).mean().item()
print(f"MSE per pixel: {mse:.2e}  |  PSNR: {10 * np.log10(1 / mse):.1f} dB")
```

## **3. Posterior Collapse**
---

The average KL of each latent dimension measures how much information it carries. Dimensions with KL ≈ 0 output the prior and are ignored by the decoder. CartPole has only two visible degrees of freedom (cart position and pole angle), so only a few active dimensions are expected.

```python
kl_per_dim = (-0.5 * (1 + logvar - mu.pow(2) - logvar.exp())).mean(dim=0).cpu().numpy()
active_dims = np.where(kl_per_dim > 0.05)[0]
print(f"Active latent dimensions (KL > 0.05): {len(active_dims)} / {len(kl_per_dim)} -> {active_dims.tolist()}")

plt.figure(figsize=(10, 4))
plt.bar(range(len(kl_per_dim)), kl_per_dim)
plt.axhline(0.05, color='r', linestyle='--')
plt.xlabel("Latent dimension")
plt.ylabel("Average KL")
plt.title("Per-dimension KL (active if > 0.05)")
plt.show()
```

## **4. Latent Traversals**
---

Starting from the mean latent of the dataset, move one active dimension at a time from -3σ to +3σ and decode. This shows which visual factor each active dimension controls.

```python
z_mean = mu.mean(dim=0)
z_std = mu.std(dim=0)
steps = np.linspace(-3, 3, 9)

fig, axes = plt.subplots(len(active_dims), len(steps), figsize=(14, 1.8 * len(active_dims)), squeeze=False)
for row, dim in enumerate(active_dims):
    z = z_mean.repeat(len(steps), 1)
    z[:, dim] += torch.tensor(steps, dtype=torch.float32, device=z.device) * z_std[dim]
    with torch.no_grad():
        frames = model.decode(z).cpu().permute(0, 2, 3, 1).numpy()
    for col in range(len(steps)):
        axes[row, col].imshow(frames[col])
        axes[row, col].set_xticks([]); axes[row, col].set_yticks([])
    axes[row, 0].set_ylabel(f"z[{dim}]", fontsize=11)
for col, s in enumerate(steps):
    axes[0, col].set_title(f"{s:+.1f}σ")
plt.tight_layout()
plt.show()
```

## **5. Physical State Probing**
---

**Question:** do the latents contain the physical state of the system, i.e. the four numbers that fully describe CartPole: cart position `x`, cart velocity `ẋ`, pole angle `θ` and angular velocity `θ̇`? If they do, a JEPA predictor working in latent space has the information it needs to model the dynamics.

### **Where the true state comes from**

The VAE dataset only stores images, so we generate new episodes here. The simulator keeps its internal physical state in `env.unwrapped.state = (x, ẋ, θ, θ̇)`. This is the same vector the standard (non-pixel) CartPole returns as its observation, and it is exactly what the renderer uses to draw each frame. For every step we save the frame together with this state, so each image is paired with its **exact ground-truth** values (not an estimate).

The episodes use the same policy as the data collection (random action held for 20 steps), so the frames come from the same distribution the VAE was trained on. Each frame is then encoded and we keep the latent mean `μ` (no random sampling), giving one 32-dimensional vector `z_t` per frame.

### **What we expect**

- **Positions** (`x`, `θ`) are visible in a single image, so they should be recoverable from `z_t` alone.
- **Velocities** (`ẋ`, `θ̇`) **cannot be seen in a single image**: a still frame doesn't show how fast things are moving. They should only be recoverable from two consecutive latents `[z_t, z_{t-1}]`, just as velocity can be computed from two positions.

```python
setup_pygame_headless()
env = create_cartpole_env()
rng = np.random.default_rng(1000)

frames, states, episode_ids, is_reset = [], [], [], []
for ep in range(40):
    obs, _ = env.reset(seed=10_000 + ep)
    reset, action = True, 0
    for t in range(300):
        frames.append(obs)
        states.append(np.array(env.unwrapped.state))
        episode_ids.append(ep)
        is_reset.append(reset)
        if t % 20 == 0:
            action = int(rng.integers(2))
        obs, _, done, _, _ = env.step(action)
        reset = done
        if done:
            obs, _ = env.reset()
env.close()

frames = torch.from_numpy(np.stack(frames)).permute(0, 3, 1, 2).float() / 255.0
states = np.stack(states)
episode_ids = np.array(episode_ids)
is_reset = np.array(is_reset)

with torch.no_grad():
    Z = model.encode(frames.to(DEVICE))[0].cpu().numpy()

STATE_NAMES = ['cart position x', 'cart velocity ẋ', 'pole angle θ', 'angular velocity θ̇']
print(f"Frames: {len(Z)}  |  mean steps between resets: {np.diff(np.where(is_reset)[0]).mean():.1f}")
```

### **State distribution**

Before measuring anything, check which states the episodes actually visit. The probes below can only be trusted inside this range: a state that never appears in the data can't be evaluated.

For reference, the cart can move in `[-2.4, 2.4]` and the episode terminates when `|θ| > 0.21` rad (12°). If the cart position histogram is concentrated near 0, the VAE has only seen the cart near the center of the screen.

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 3))
for i, name in enumerate(STATE_NAMES):
    axes[i].hist(states[:, i], bins=50)
    axes[i].set_title(name)
plt.tight_layout()
plt.show()
```

### **Linear probes**

A **probe** is a simple model trained to read a quantity out of a representation. Here we fit one **linear regression** per state variable:

$$\hat{s}_i = w_i^\top z_t + b_i$$

where `s_i` is one of `x`, `ẋ`, `θ`, `θ̇`. The VAE is frozen; only the weights `w_i, b_i` are learned.

**Why linear?** A linear model can't create information that isn't there; it can only find it if the latent already encodes it in a simple, direct way. A high score therefore means the variable is *explicitly* present in the latent, which is what makes it easy for the JEPA predictor to use. (A powerful nonlinear probe could partly compute the answer itself, which would make the test less informative.)

**Details:**
- We use **ridge regression** (`sklearn.linear_model.Ridge`), which is linear regression with a small penalty on large weights to avoid overfitting.
- **Train/test split by episode:** the probe is fitted on episodes 0–29 and evaluated on episodes 30–39. Consecutive frames are almost identical, so splitting frame-by-frame would let the probe "memorize" neighbors of the test frames; splitting by episode tests on genuinely unseen trajectories.
- **Score: R²** (coefficient of determination) on the test episodes. R² = 1 means perfect prediction; R² = 0 means the probe is no better than always predicting the average value. For example, R² = 0.75 means the probe explains 75% of the variance of that variable.

We run the probe twice:
1. **From `z_t`** (32 features): what a single frame's latent contains.
2. **From `[z_t, z_{t-1}]`** (64 features, the current and previous latents concatenated): what two consecutive frames contain together. The first frame after each reset is skipped, because its "previous frame" belongs to a different episode.

**How to read it:** positions should score close to 1 in both cases. Velocities should score clearly higher with two frames. Any velocity score from a single frame comes from correlations created by the policy (e.g. when the pole is already tilted, it is usually also falling), not from the image itself.

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score

train = episode_ids < 30
test = ~train

def probe(features, targets, train_mask, test_mask):
    preds = np.zeros((test_mask.sum(), targets.shape[1]))
    for i in range(targets.shape[1]):
        reg = Ridge(alpha=1.0).fit(features[train_mask], targets[train_mask, i])
        preds[:, i] = reg.predict(features[test_mask])
    r2 = [r2_score(targets[test_mask, i], preds[:, i]) for i in range(targets.shape[1])]
    return r2, preds

# Single frame: z_t
r2_single, preds_single = probe(Z, states, train, test)

# Two consecutive frames: [z_t, z_{t-1}] (skip the first frame after each reset)
idx = np.where(~is_reset)[0]
Z_pair = np.concatenate([Z[idx], Z[idx - 1]], axis=1)
r2_pair, _ = probe(Z_pair, states[idx], train[idx], test[idx])

print(f"{'State variable':22s} {'R² from z_t':>12s} {'R² from [z_t, z_t-1]':>22s}")
for name, a, b in zip(STATE_NAMES, r2_single, r2_pair):
    print(f"{name:22s} {a:12.3f} {b:22.3f}")
```

**Predicted vs. true values.** Each point is one test frame: its true value (horizontal axis) against the value the single-frame probe predicts (vertical axis). Points on the red diagonal are perfect predictions. A tight band along the diagonal means the variable is encoded accurately; a wide cloud means the probe is mostly guessing.

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, name in enumerate(STATE_NAMES):
    true = states[test, i]
    axes[i].scatter(true, preds_single[:, i], s=2, alpha=0.3)
    lims = [true.min(), true.max()]
    axes[i].plot(lims, lims, 'r--')
    axes[i].set_title(f"{name}\nR² = {r2_single[i]:.3f}")
    axes[i].set_xlabel("true")
axes[0].set_ylabel("predicted from z_t")
plt.tight_layout()
plt.show()
```

### **Latent trajectory**

A visual check of how the latent space is organized. **PCA** (principal component analysis) finds the two directions along which the 32-dimensional latents vary the most and projects every test frame onto them, giving a 2D picture of the latent space.

Each point is one frame, colored by its true pole angle (left) or cart position (right). If the colors change **smoothly** across the plot (a gradient rather than random speckles), nearby latents correspond to similar physical states. This is what a predictor needs: small changes in the state give small, predictable changes in the latent.

```python
from sklearn.decomposition import PCA

Z_2d = PCA(n_components=2).fit_transform(Z[test])
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, i in zip(axes, [2, 0]):
    sc = ax.scatter(Z_2d[:, 0], Z_2d[:, 1], c=states[test, i], cmap='coolwarm', s=3)
    plt.colorbar(sc, ax=ax, label=STATE_NAMES[i])
    ax.set_xlabel("PC 1"); ax.set_ylabel("PC 2")
axes[0].set_title("Latents colored by pole angle")
axes[1].set_title("Latents colored by cart position")
plt.tight_layout()
plt.show()
```

## **How to Read the Results**
---

- **Reconstruction**: PSNR above ~35 dB means the cart and pole are reconstructed precisely.
- **Active dimensions**: a handful (≈2–4) is expected for CartPole; the remaining dimensions collapse to the prior, which is harmless for the JEPA as long as the probes below are good.
- **Probes from `z_t`**: R² close to 1 for `x` and `θ` means the latent encodes the visible state linearly. Velocity R² from a single frame is limited (any correlation comes from the data-collection policy, not from the image).
- **Probes from `[z_t, z_{t-1}]`**: R² close to 1 for the velocities means the dynamics are recoverable from consecutive latents. The JEPA predictor should therefore receive more than one frame (or a recurrent state, as the MDN-RNN in World Models).
- **State distribution**: if the cart position only covers a small part of `[-2.4, 2.4]`, the VAE has never seen the cart near the edges, and its latents may not generalize there.
